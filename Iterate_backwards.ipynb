```python
import numpy as np
```

```python
end = 20
start = 6
step_size = 3
max_iterations = (end-start) // step_size +1


A = np.zeros(max_iterations)
# print(max_iterations, np.shape(A)[0])

for MCS in range(end, start, -step_size):
    # print(MCS, (MCS-start)//step_size-1)
    A[(MCS-start)//step_size] = MCS
print(A)
```
